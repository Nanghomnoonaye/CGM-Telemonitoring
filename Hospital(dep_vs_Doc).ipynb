```python
import scipy.stats as stats

# Convert p-value to Z-score
p = 0.410
z = stats.norm.ppf(1 - p / 2)

print(f"p = {p:.3f} | Z = {z:.4f}")
```

```python
# H0 : no RS
# Ha : RS

#P is much bigger than 0.05.
 # we did not find significant results,we cannot ​support our alternative hypothesis, and therefore stick with our null hypothesis.
```
