# M5-V2 ANOVA analysis

```python
# Y -> moisture percentages (numeric)
# X -> matchine. (categoric)

need to perform ANOVA, just to see if machine is a significant influence factor
#we will study if the moisture percentage is equal for ​each machine.
```

```python
one-way ANOVA. -> graph -> individual value plot
                  option -> uncheck (assume =)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data matching the plot points
np.random.seed(42)
data = [
    [9.5, 10.9, 11.5, 11.9, 12.3, 13.8, 14.5],  # Machine 1
    [6.6, 7.1, 8.8, 9.3, 11.0, 14.1],  # Machine 2
    [8.4, 8.8, 10.6, 11.1, 11.3, 12.9, 13.0, 13.7],  # Machine 3
    [8.8, 9.3, 9.7, 10.2, 12.3],  # Machine 4
]

x_coords = [1, 2, 3, 4]
means = [np.mean(d) for d in data]

fig, ax = plt.subplots(figsize=(6, 3.5))

# Individual points
for x, pts in zip(x_coords, data):
    ax.scatter([x] * len(pts), pts, color='gray', s=30, alpha=0.8, zorder=2)

# Mean line and points
ax.plot(
    x_coords,
    means,
    color='#1f77b4',
    marker='o',
    markersize=5,
    lw=1,
    zorder=3,
)

# Red circles highlight on Machine 1 and Machine 3
ax.scatter(
    [1, 3],
    [means[0], means[2]],
    s=700,
    facecolors='none',
    edgecolors='#a61c3c',
    linewidth=2,
    zorder=4,
)

# Styling
ax.set_title(
    'Individual Value Plot of Moisture vs Machine', fontsize=11, pad=8
)
ax.set_ylabel('Moisture', fontweight='bold', fontsize=10)
ax.set_xticks(x_coords)
ax.set_xticklabels(
    ['Machine 1', 'Machine 2', 'Machine 3', 'Machine 4'], fontsize=9
)
ax.set_yticks(range(6, 16))
ax.set_ylim(5.8, 15.2)

# Clear background & simple frame border
ax.set_facecolor('none')
for spine in ax.spines.values():
    spine.set_color('#888888')

plt.tight_layout()
plt.show()
```

```python
The Individual Value Plot shows the measured moisture percentage for ​each machine.

The blue line connects the group means.
# On first sight, ​we see differences in the group means, as the line is not horizontal.

# Machine 1 and 3
appear to produce coffee with relatively high moisture content.
Now that, you will never get exactly horizontal lines ​even if the group means are in reality equal. ​
Especially for small data sets, you should statistically test ​whether the differences between the means are real or due to random fluctuations. ​

# p-value indicates the chance that such difference ​occur due to random fluctuation.
p-value ​only tells you where an influence factor has a significant effect.

```

```python
That is why, we look at the statistical analysis (p-value)
   indicates the chance that such difference ​occur due to random fluctuation.
```

In this example, it is 1.2%. ​As the probability that this difference occurred by chance is so ​low, we conclude that the machines differ. ​In fact, the threshold is often sets to 5% and ​the p-value here is below this 5%. ​This means that the effect is significant and that it translates to the population. ​If the p-value would have been bigger that 0.05, it means you either ​did not find the real difference or you did not gather enough data to prove it. ​

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(3, 0.5))

cols = [
    (0.15, 'R-sq', '26.11%'),
    (0.5, 'R-sq(adj)', '19.95%'),
    (0.85, 'R-sq(pred)', '8.77%'),
]  # cite: 7
for x, h, v in cols:
    ax.text(x, 0.7, h, weight='bold', ha='center')  # cite: 7
    ax.text(x, 0.1, v, ha='center')  # cite: 7

ax.axhline(0.45, color='#ddd', lw=1)
ax.axis('off')
plt.show()
```

```python
To see how relevant and influence factor is, we can take a look at the R-squared. ​

R-squared Always lies between zero and 100% and tells you ​how much of the variation in your Y is explained by X variable. ​

#R squared is 26% so the influence factor machine accounts for ​or
     explains 26% of the variation in the CTQ moisture percentage. ​
```

# M5-V3 ANOVA residual analysis

Screenshot 2026-09-20 at 10.55.43 pm.png

ANOVA analysis gave us a p-value, which shows a statistical significant ​difference between the average moisture percentages of the machines, ​because the p-value is below 0.05. ​This difference in the means can also be seen in the individual value plot, ​as the line connecting the means is not horizontal. ​Let's take a look at the R squared. ​This shows that the influence factor machine explains 26% of ​the variation in the moisture percentage. ​However, before we can completely trust these conclusions, ​we have to validate the assumptions underlying the ANOVA. ​These checks are called the residual analysis, and ​this is the last and final step of your ANO

```python
                                           To validate the assumptions,

we will check if the residuals are normally ​distributed and if there are any outliers or other irregularities present.
```

```python
import matplotlib.pyplot as plt

pts = [9.5, 10.9, 11.5, 11.9, 12.3, 13.8, 14.5]  # cite: 7
mean = 11.9  # cite: 7

fig, ax = plt.subplots(figsize=(5, 3))

# Plot data
ax.scatter([1] * 7, pts, color='gray', zorder=2)  # cite: 7
ax.scatter(1, mean, color='#1f77b4', zorder=3)  # cite: 7
ax.plot([1, 1.8], [mean, 10.5], color='#1f77b4', lw=1)  # cite: 7

# Annotations
el = dict(boxstyle='round,pad=0.3', fc='none', ec='#555', lw=1.5)
arr = dict(arrowstyle='->', color='#555', lw=1.5)

ax.text(2.2, 12, 'Estimated\nmean', ha='center', bbox=el)  # cite: 7
ax.annotate('', xy=(1.05, mean), xytext=(1.8, 12), arrowprops=arr)  # cite: 7

ax.text(2.2, 7.5, 'Measurement', ha='center', bbox=el)  # cite: 7
ax.annotate('', xy=(1.05, 9.5), xytext=(1.8, 7.5), arrowprops=arr)  # cite: 7

# Curly brace representation
ax.text(1.2, 10.7, '}', fontsize=28, color='#800020', va='center')  # cite: 7

ax.set(
    xlim=(0.5, 2.8),
    ylim=(5.5, 15.5),
    ylabel='Moisture',
    xticks=[1],
    xticklabels=['Machine 1'],
)  # cite: 7
ax.set_title('Individual Value', loc='left', fontsize=9)  # cite: 7

plt.tight_layout()
plt.show()
```

```python
Every dot in the graph is one measurement.

We also know the value that we would expect from a measurement for Machine 1. ​That is the estimated mean.

     So there is a difference between the measurement and our expectation.
     This difference is not explained by our influence factor machine. ​It is left over variation, called the residual.

#The residuals are calculated by subtracting the expected value from ​each observation.

Stat > ANOVA > One Way > Residual -> 4 in one
```

```python
#Remember, that we needed to check two things in the residual analysis.

# normality assumption. --->  can be checked in the probability plot
Q: Are your residuals normally distributed?
     1st gragh ---> ​Yes, they are.

Q: there are no outliers or ​irregularities in the residuals.
​To check this assumption, we take a look at the fourth graph (Observation order).
 # Ans= We see that there are no outliers or strange patterns present. ​
 This means that this assumption is also satisfied, and ​that the original analysis is valid.
```
