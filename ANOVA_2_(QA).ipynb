# M5-V4 Kruskal-Wallis test

To interpret the output, you have to look at the medians for each group and ​of course, add the P-value to see if this difference is statistically significant.

```python
numerical Y and a categorical X

# eg. call center
-> some receive Training
--> some do not receive Training

# Test
whether this training has any effect on the total handling time, ​
that is, the total time it takes an employee to handle an incoming call.

Y = Total handling time (numerical)
X = training or no training (categorical)
```

from ANOVA residual analysis

```python
As you can see, the residuals are not normally distributed.
​Furthermore, the time graph also shows some irregularities,

​hence we need to apply an alternative analysis technique.

# ​In this case, we will apply the Kruskal-Wallis test. ​
```

```python
Minitab -> Stat -> Nonparametrics -> Kruskal-Wallis
```

```python
Kruskal-Wallis analysis is based on # medians
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 1.2))

# Data
cols = [
    (0.05, 'Training', ['No', 'Yes', 'Overall'], 'left'),  # cite: 8
    (0.25, 'N', ['35', '21', '56'], 'center'),  # cite: 8
    (0.48, 'Median', ['246', '191', ''], 'center'),  # cite: 8
    (0.72, 'Mean Rank', ['33.9', '19.5', '28.5'], 'center'),  # cite: 8
    (0.95, 'Z-Value', ['3.18', '-3.18', ''], 'right'),  # cite: 8
]

# Render
for x, head, vals, align in cols:
    ax.text(x, 0.82, head, weight='bold', ha=align)  # cite: 8
    for i, v in enumerate(vals):
        ax.text(x, 0.58 - i * 0.24, v, ha=align)  # cite: 8

# Divider line
ax.axhline(0.72, color='#eee', lw=1)
ax.axis('off')

plt.tight_layout()
plt.show()
```

The median handling time for employees with training is 191 seconds. ​And the median handling time for employees without training is 246 seconds. ​So it took people without training a lot more time to ​handle a phone call than it did for people that received the training. ​The difference is nearly a minute.

​The p-value is also given.
​It is lower than 0.05. ​So the difference in medians that we found is statistically significant. ​It can be concluded that the training is an effective ​method to lower the total handling time by 55 seconds

### an example from agriculture

```python
we are growing tomatoes, and we wish to maximize the yield
and that is the total amount of kg of tomatoes that we are harvest.

For this, you experiment with two different fertilizers. ​we call them fertilizer A and fertilizer B.

# Q: ​Which fertilizer produces more tomatoes than the other? ​

In order to answer this question, we set up a small experiment.

-> A possible way is to take a field and ​divide it into 20 pieces, and plant tomatoes here.
​Then select ten pieces randomly and put fertilizer A on them, fertilizer B on the other ten fields, or sub fields.
​Next you wait until tomatoes are grown.

​You harvest them and you measure the yield for each sub field. ​And this could be the data that you get.
```

```python
Minitab -> Stat -> 2-sample -> each sample in its own column

look at
# Mean. # P-value

 # StDev (variation)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5.5, 1.3))

cols = [
    (0.05, 'Sample', ['Fertilizer A', 'Fertilizer B'], 'left'),  # cite: 9
    (0.32, 'N', ['10', '10'], 'center'),  # cite: 9
    (0.48, 'Mean', ['5.66', '8.40'], 'center'),  # cite: 9
    (0.68, 'StDev', ['2.05', '2.46'], 'center'),  # cite: 9
    (0.92, 'SE Mean', ['0.65', '0.78'], 'right'),  # cite: 9
]

for x, head, vals, align in cols:
    ax.text(x, 0.82, head, weight='bold', ha=align)  # cite: 9
    for i, v in enumerate(vals):
        ax.text(x, 0.54 - i * 0.28, v, ha=align)  # cite: 9

ax.axhline(0.72, color='#eee', lw=1)
ax.text(0.05, 0.02, 'P-Value = 0.015', weight='bold', ha='left')
ax.axis('off')

plt.tight_layout()
plt.show()
```

```python
The estimated difference between the mean of fertilizer A and B is 2.74.

p-value is 0.015. In this example, this probability is 1.5%, it's relatively small.
```

So given this data of 20 fields, we can conclude that there's ​only a slight chance of 1.5% that the influence factor ​fertilizer has no effect on yield and that the means are equal.

Hence, if we put it in more normal English, ​the p-value is smaller than 0.05 and ​we conclude that the fertilizer has a significant effect on the yield.

Now, to actually get some knowledge out of this, ​we have to actually go back to our data and look at the mean levels. ​Because if we now know that the fertilizers are different, ​which is the one that we have to use? ​Well, that's obviously the one with the highest mean, and that's fertilizer B.

# M5-V6 Test for equality of variances

```python
# eg

four machines producing coffee
We investigated that the average moisture content ​of the coffee beans depended on the machine it was produced on.
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = [
    [9.5, 10.9, 11.5, 11.9, 12.3, 13.8, 14.5],  # Machine 1
    [6.6, 7.1, 8.8, 9.3, 11.0, 14.1],  # Machine 2
    [8.4, 8.8, 10.6, 11.1, 11.3, 12.9, 13.0, 13.7],  # Machine 3
    [8.8, 9.3, 9.7, 10.2, 12.3],  # Machine 4
]
x_coords, means = [1, 2, 3, 4], [np.mean(d) for d in data]

fig, axes = plt.subplots(1, 3, figsize=(14, 3.5), sharey=True)

for i, ax in enumerate(axes):
    # Data & Mean Line
    for x, pts in zip(x_coords, data):
        ax.scatter([x] * len(pts), pts, color='gray', s=25, zorder=2)
    ax.plot(
        x_coords,
        means,
        color='#1f77b4',
        marker='o',
        ms=4,
        lw=1,
        zorder=3,
    )

    # Subplot Specific Annotations
    if i == 0:  # Fig 1: Highlight all 4 means
        ax.scatter(
            x_coords,
            means,
            s=500,
            fc='none',
            ec='#a61c3c',
            lw=2,
            zorder=4,
        )  # cite: 10
    elif i == 1:  # Fig 2: Vertical double arrows for variance
        ax.annotate(
            '',
            xy=(2.2, 14.3),
            xytext=(2.2, 5.8),
            arrowprops=dict(
                arrowstyle='<->', color='#a61c3c', lw=2, mutation_scale=12
            ),
        )  # cite: 11
        ax.annotate(
            '',
            xy=(4.2, 12.2),
            xytext=(4.2, 7.8),
            arrowprops=dict(
                arrowstyle='<->', color='#a61c3c', lw=2, mutation_scale=12
            ),
        )  # cite: 11
    elif i == 2:  # Fig 3: Highlight Machine 2 & 4 means + side arrow
        ax.scatter(
            [2, 4],
            [means[1], means[3]],
            s=500,
            fc='none',
            ec='#a61c3c',
            lw=2,
            zorder=4,
        )  # cite: 12
        ax.annotate(
            '',
            xy=(0.4, 9.7),
            xytext=(-0.1, 9.7),
            arrowprops=dict(
                arrowstyle='->', color='#a61c3c', lw=2, mutation_scale=12
            ),
        )  # cite: 12

    # Formatting
    ax.set_title(
        'Individual Value Plot of Moisture vs Machine', fontsize=9, pad=6
    )  # cite: 10, 11, 12
    ax.set_xlabel('Machine', fontweight='bold', fontsize=9)  # cite: 10, 11, 12
    ax.set_xticks(x_coords)
    ax.set_xticklabels(
        ['Machine 1', 'Machine 2', 'Machine 3', 'Machine 4'], fontsize=8
    )  # cite: 10, 11, 12
    ax.set_facecolor('none')
    for spine in ax.spines.values():
        spine.set_color('#888888')

axes[0].set_ylabel('Moisture', fontweight='bold', fontsize=9)  # cite: 10
axes[0].set_yticks(range(6, 16))
axes[0].set_ylim(5.5, 15.2)

plt.tight_layout()
plt.show()
```

```python
#1 -> means of moisture % differ across the four machines.

         But what about the variation within a machine?

#2 -> Machine two has a wide range of moisture contents, ​while machine four appeared a lot more consistent. ​

-> For producing consistent coffee, ​we probably want the machine with the smallest variation. ​

#3 -> The mean level is near equal for machine two and four.
However, the variation of the two machines may differ.
```

```python
import matplotlib.gridspec as gridspec
import matplotlib.pyplot as plt

# Data
data = [
    [9.5, 10.9, 11.5, 11.9, 12.3, 13.8, 14.5],  # Machine 1
    [9.4, 9.6, 6.6, 9.3, 14.1, 9.5, 11.0, 7.1, 11.0, 8.8],  # Machine 2 ts
    [8.4, 8.8, 10.6, 11.1, 11.3, 12.9, 13.0, 13.7],  # Machine 3
    [8.8, 9.5, 12.3, 10.0, 8.8, 9.5, 10.2, 8.8, 10.1, 9.3],  # Machine 4 ts
]

fig = plt.figure(figsize=(9, 4.5))
gs = gridspec.GridSpec(2, 2, width_ratios=[1.2, 1], hspace=0.4, wspace=0.3)

ax_left = fig.add_subplot(gs[:, 0])
ax_rt1 = fig.add_subplot(gs[0, 1])
ax_rt2 = fig.add_subplot(gs[1, 1])

# Left: Individual Value Plot
x_coords = [1, 2, 3, 4]
means = [
    sum(d) / len(d) for d in [data[0], data[1], data[2], data[3]]
]  # cite: 13
for x, pts in zip(x_coords, data):
    ax_left.scatter([x] * len(pts), pts, color='gray', s=25, zorder=2)
ax_left.plot(x_coords, means, color='#1f77b4', marker='o', ms=4, lw=1, zorder=3)
ax_left.annotate(
    '',
    xy=(2, 14.3),
    xytext=(2, 5.8),
    arrowprops=dict(arrowstyle='<->', color='#a61c3c', lw=2),
)  # cite: 13
ax_left.annotate(
    '',
    xy=(4, 12.2),
    xytext=(4, 7.8),
    arrowprops=dict(arrowstyle='<->', color='#a61c3c', lw=2),
)  # cite: 13
ax_left.set_title(
    'Individual Value Plot of Moisture vs Machine', fontsize=8
)  # cite: 13
ax_left.set(
    ylabel='Moisture',
    xticks=x_coords,
    xticklabels=['Machine 1', 'Machine 2', 'Machine 3', 'Machine 4'],
    ylim=(5.5, 15.5),
)  # cite: 13

# Right Top & Bottom: Time Series
for ax, d, title, ylbl in [
    (
        ax_rt1,
        data[1],
        'Time Series Plot of Moisture machine 2',
        'Moisture machine 2',
    ),  # cite: 13
    (
        ax_rt2,
        data[3],
        'Time Series Plot of Moisture machine 4',
        'Moisture Machine 4',
    ),
]:  # cite: 13
    ax.plot(
        range(1, 11),
        d,
        color='#1f77b4',
        marker='o',
        ms=3,
        lw=0.8,
    )  # cite: 13
    ax.set_title(title, fontsize=8)  # cite: 13
    ax.set(
        ylabel=ylbl,
        xlabel='Index',
        xticks=range(1, 11),
        ylim=(5.5, 15.5),
    )  # cite: 13

# Arrow connectors from Left to Right
arrow_box = dict(
    boxstyle='rarrow,pad=0.2', fc='#c2bbf0', ec='#6c5ce7', lw=1
)  # cite: 13
ax_left.text(
    2.2, 12.5, '   ', bbox=arrow_box, rotation=-10, transform=ax_left.transData
)  # cite: 13

for ax in [ax_left, ax_rt1, ax_rt2]:
    ax.set_facecolor('none')
    for s in ax.spines.values():
        s.set_color('#888888')

fig.suptitle(
    'Test for equal variances\nExample – Moisture %',
    x=0.02,
    y=0.98,
    ha='left',
    fontsize=11,
    fontweight='bold',
)  # cite: 13
plt.tight_layout()
plt.show()
```

```python
#4 The moisture content of machine two goes up and down more than that of machine four.

If the machine produces very inconsistently, ​meaning a lot of variation, ​this might cause problems. ​
# could lead to a lot of scrap and defects which will cost you money.
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

# Data
d2 = [9.4, 9.6, 6.6, 9.3, 14.1, 9.5, 11.0, 7.1, 11.0, 8.8]  # cite: 13
d4 = [8.8, 9.5, 12.3, 10.0, 8.8, 9.5, 10.2, 8.8, 10.1, 9.3]  # cite: 13
x = range(1, 11)  # cite: 13, 14

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6, 5), sharex=True)

# Plot 1: Machine 2
ax1.plot(x, d2, color='#1f77b4', marker='o', ms=3, lw=0.8)  # cite: 13
ax1.axhline(10.5, color='#a61c3c', lw=2.5)  # cite: 13
ax1.add_patch(
    Ellipse(
        (6, 13), 7.5, 4.5, fill=False, edgecolor='#a61c3c', lw=2.5
    )  # cite: 13
)
ax1.set_title('Time Series Plot of Moisture machine 2', fontsize=9)  # cite: 13
ax1.set(ylabel='Moisture machine 2', ylim=(5.5, 15.5))  # cite: 13

# Plot 2: Machine 4
ax2.plot(x, d4, color='#1f77b4', marker='o', ms=3, lw=0.8)  # cite: 13, 14
ax2.axhline(10.5, color='#a61c3c', lw=2.5)  # cite: 14
ax2.scatter(
    3, 12.3, s=400, facecolors='none', edgecolors='#a61c3c', lw=2.5
)  # cite: 14
ax2.set_title('Time Series Plot of Moisture machine 4', fontsize=9)  # cite: 14
ax2.set(
    ylabel='Moisture Machine 4', xlabel='Index', xticks=x, ylim=(5.5, 15.5)
)  # cite: 14

for ax in (ax1, ax2):
    ax.set_facecolor('none')
    for s in ax.spines.values():
        s.set_color('#888888')

plt.tight_layout()
plt.show()
```

```python
#eg
If you want your moisture % to be lower than 10.5%

# in machine 2
there is a lot of variation, ​a lot of your batches will show a moisture percentage which is higher than this 10.5
---> those batches will have to be thrown away or reprocessed.

# in machine 4
the variation is smaller, ​this will happen less often.

----> ​Therefore, you will probably like to know if one ​of the machine produces more consistently than the others.
```

The output of the ANOVA analysis already gave us some information on the variation. ​We see that machine four has a smaller standard deviation and therefore, ​produces most consistently within the sample at least. ​However, are these differences between ​these standard deviation is statistically significant, ​or is it a coincidence that machine four has the smallest variation? ​

```python
Minitab -> stat -> ANOVA -> test for equal variances
```

Screenshot 2026-09-21 at 10.28.42 pm.png

The estimates for the standard deviation in ​moisture percentage for the machines are given. ​To check if the differences between these are statistically significant, ​we look at the p-value, ​compute it according to Levene's method. ​In this case, the p-value is relatively large. ​It shows us that there is a 32.5 percent ​chance that the difference between the machines is a chance fluctuation, ​which is much larger than our threshold of five percent. ​This means that we did not find evidence that ​there is a difference in variation between machines, ​and we cannot draw any conclusions on what the machine produces most consistently.
